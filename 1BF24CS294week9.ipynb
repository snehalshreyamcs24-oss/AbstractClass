```python

# Week 9: Resolution

def negate(literal):
    if literal.startswith("~"):
        return literal[1:]
    return "~" + literal


def resolve(c1, c2):
    resolvents = set()

    for literal in c1:
        opposite = negate(literal)

        if opposite in c2:
            new_clause = (c1 - {literal}) | (c2 - {opposite})

            # Remove tautological clauses
            if not any(negate(x) in new_clause
                       for x in new_clause):
                resolvents.add(frozenset(new_clause))

    return resolvents


def resolution(kb, query):
    clauses = set(kb)

    # Negate the query
    clauses.add(frozenset({negate(query)}))

    while True:
        new_clauses = set()
        clause_list = list(clauses)

        for i in range(len(clause_list)):
            for j in range(i + 1, len(clause_list)):
                resolvents = resolve(
                    clause_list[i], clause_list[j]
                )

                if frozenset() in resolvents:
                    return True

                new_clauses.update(resolvents)

        if new_clauses.issubset(clauses):
            return False

        clauses.update(new_clauses)


# Knowledge Base in CNF
kb = {
    # John likes all food
    frozenset({"~Food(Peanuts)", "Likes(John,Peanuts)"}),

    # Eats(x,y) AND NotKilled(x) -> Food(y)
    frozenset({
        "~Eats(Anil,Peanuts)",
        "~NotKilled(Anil)",
        "Food(Peanuts)"
    }),

    # Alive(x) -> NotKilled(x)
    frozenset({
        "~Alive(Anil)",
        "NotKilled(Anil)"
    }),

    # Facts
    frozenset({"Eats(Anil,Peanuts)"}),
    frozenset({"Alive(Anil)"})
}

query = "Likes(John,Peanuts)"

if resolution(kb, query):
    print("Query proved by Resolution")
else:
    print("Query not proved")
```
