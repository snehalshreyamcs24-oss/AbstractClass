```python

# Week 8: FOL Forward Chaining

facts = {
    "American(Robert)",
    "Missile(T1)",
    "Owns(A,T1)",
    "Enemy(A,America)"
}

rules = [
    ({"Missile(T1)"}, "Weapon(T1)"),
    ({"Enemy(A,America)"}, "Hostile(A)"),
    ({"Missile(T1)", "Owns(A,T1)"}, "Sells(Robert,T1,A)"),
    ({"American(Robert)", "Weapon(T1)",
      "Sells(Robert,T1,A)", "Hostile(A)"}, "Criminal(Robert)")
]

goal = "Criminal(Robert)"

while True:
    new_fact = False

    for premises, conclusion in rules:
        if premises.issubset(facts) and conclusion not in facts:
            facts.add(conclusion)
            print("Inferred:", conclusion)
            new_fact = True

    if goal in facts:
        print("Goal proved:", goal)
        break

    if not new_fact:
        print("Goal not proved")
        break
```
