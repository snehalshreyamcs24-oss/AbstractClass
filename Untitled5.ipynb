```python
import random

def objective(x):
    return -(x - 5) ** 2 + 25

def hill_climbing():
    current = random.randint(0, 10)

    while True:
        neighbors = [current - 1, current + 1]

        neighbors = [n for n in neighbors if 0 <= n <= 10]

        best_neighbor = max(neighbors, key=objective)

        if objective(best_neighbor) <= objective(current):
            break

        current = best_neighbor

    return current, objective(current)

solution, value = hill_climbing()

print("Best solution:", solution)
print("Maximum value:", value)
```

```python
import random
import math

def f(x):
    return -(x - 5)**2 + 25

def simulated_annealing():
    current = 0
    temperature = 100
    cooling_rate = 0.95

    while temperature > 0.1:
        new = current + random.choice([-1, 1])

        delta = f(new) - f(current)

        if delta > 0:
            current = new
        else:
            probability = math.exp(delta / temperature)

            if random.random() < probability:
                current = new

        temperature *= cooling_rate

    return current, f(current)

x, value = simulated_annealing()

print("Best solution:", x)
print("Maximum value:", value)
```
