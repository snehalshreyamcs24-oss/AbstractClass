```python

# Week 10: Alpha-Beta Pruning

def alpha_beta(node, alpha, beta, maximizing, path="Root"):

    # Terminal node
    if isinstance(node, int):
        return node, path

    if maximizing:
        best_value = float('-inf')
        best_path = ""

        for i, child in enumerate(node):
            value, child_path = alpha_beta(
                child, alpha, beta, False,
                path + " -> " + str(i + 1)
            )

            if value > best_value:
                best_value = value
                best_path = child_path

            alpha = max(alpha, best_value)

            if alpha >= beta:
                remaining = list(range(i + 2, len(node) + 1))
                if remaining:
                    print("Pruned at", path,
                          "children:", remaining)
                break

        return best_value, best_path

    else:
        best_value = float('inf')
        best_path = ""

        for i, child in enumerate(node):
            value, child_path = alpha_beta(
                child, alpha, beta, True,
                path + " -> " + str(i + 1)
            )

            if value < best_value:
                best_value = value
                best_path = child_path

            beta = min(beta, best_value)

            if beta <= alpha:
                remaining = list(range(i + 2, len(node) + 1))
                if remaining:
                    print("Pruned at", path,
                          "children:", remaining)
                break

        return best_value, best_path


# Example game tree
tree = [
    [3, 5, 6],
    [2, 9, 1],
    [7, 5, 4]
]

value, path = alpha_beta(
    tree, float('-inf'), float('inf'), True
)

print("Root value:", value)
print("Optimal path:", path)
```
