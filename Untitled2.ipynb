```python
import heapq


# Calculate Manhattan distance for a puzzle state
def manhattan_distance(state, goal):
    distance = 0

    for tile in range(1, 9):
        # Position of tile in current state
        current_pos = state.index(tile)

        # Position of tile in goal state
        goal_pos = goal.index(tile)

        current_row = current_pos // 3
        current_col = current_pos % 3

        goal_row = goal_pos // 3
        goal_col = goal_pos % 3

        distance += abs(current_row - goal_row)
        distance += abs(current_col - goal_col)

    return distance


# Generate possible next states
def get_neighbors(state):
    neighbors = []

    zero_pos = state.index(0)

    row = zero_pos // 3
    col = zero_pos % 3

    # Up, Down, Left, Right
    moves = [
        (-1, 0),
        (1, 0),
        (0, -1),
        (0, 1)
    ]

    for dr, dc in moves:
        new_row = row + dr
        new_col = col + dc

        if 0 <= new_row < 3 and 0 <= new_col < 3:

            new_pos = new_row * 3 + new_col

            new_state = list(state)

            # Swap blank with tile
            new_state[zero_pos], new_state[new_pos] = \
                new_state[new_pos], new_state[zero_pos]

            neighbors.append(tuple(new_state))

    return neighbors


# A* Algorithm
def a_star(initial, goal):

    # Priority queue: (f, g, state)
    open_list = []

    g_cost = {initial: 0}

    parent = {initial: None}

    h = manhattan_distance(initial, goal)

    heapq.heappush(
        open_list,
        (h, 0, initial)
    )

    visited = set()

    while open_list:

        f, g, current = heapq.heappop(open_list)

        if current in visited:
            continue

        visited.add(current)

        # Goal reached
        if current == goal:

            path = []

            while current is not None:
                path.append(current)
                current = parent[current]

            return path[::-1]

        # Generate next states
        for neighbor in get_neighbors(current):

            if neighbor in visited:
                continue

            new_g = g + 1

            if neighbor not in g_cost or new_g < g_cost[neighbor]:

                g_cost[neighbor] = new_g

                h = manhattan_distance(neighbor, goal)

                f = new_g + h

                parent[neighbor] = current

                heapq.heappush(
                    open_list,
                    (f, new_g, neighbor)
                )

    return None


# Print puzzle state
def print_state(state):

    for i in range(0, 9, 3):
        print(state[i:i + 3])

    print()


# -------------------------
# Main Program
# -------------------------

initial = (
    1, 2, 3,
    0, 4, 6,
    7, 5, 8
)

goal = (
    1, 2, 3,
    4, 5, 6,
    7, 8, 0
)

print("Initial State:")
print_state(initial)

print("Goal State:")
print_state(goal)

# Find solution
solution = a_star(initial, goal)

if solution:

    print("Solution Found!")
    print("Number of Moves:", len(solution) - 1)
    print()

    for i, state in enumerate(solution):

        print("Step", i)
        print_state(state)

else:
    print("No solution exists.")
print("Initial State:")
print_state(initial)

print("Goal State:")
print_state(goal)

# Calculate initial heuristic
initial_h = manhattan_distance(initial, goal)
initial_g = 0
initial_f = initial_g + initial_h

print("Initial g(n):", initial_g)
print("Initial h(n):", initial_h)
print("Initial f(n):", initial_f)
print()

# Find solution
solution = a_star(initial, goal)

if solution:

    print("Solution Found!")
    print("Number of Moves:", len(solution) - 1)
    print()

    for i, state in enumerate(solution):

        print("Step", i)
        print_state(state)

else:
    print("No solution exists.")
```
